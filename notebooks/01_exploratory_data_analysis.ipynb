Tworzę macierz konfliktów -> segregation of duties 

In SAP there is four-eyes-rule, which determines that single worker can not have "acces to two responsibliities" !!!!! formalize it 

Wszystkie nazwy kodów transakcji (**T-Codes**) oraz struktura macierzy konfliktów (**SoD Rules**) bazują na oficjalnej nomenklaturze systemu **SAP ERP** oraz standardach audytowych:

* **T-Codes (np. `FK01`, `ME21N`, `MIRO`):** Stanowią oficjalne skróty operacji w modułach SAP financial accounting oraz materials management, zgodne ze specyfikacją [SAP Help Portal - Transaction Codes](https://help.sap.com/docs/SAP_ACCESS_CONTROL).
* **Macierz Ryzyk SoD (`SOD_01`, `CRITICAL`):** Odzwierciedla strukturę reguł wykorzystywaną w modułach klasy GRC (Governance, Risk, and Compliance) — w tym *SAP GRC Access Control Rulebook* oraz wytycznych organizacji **ISACA** dotyczących kontroli wewnętrznych ([ISACA Segregation of Duties Framework](https://www.isaca.org)).

```python
sod_rules = [
    {
        "risk_id": "SOD_01",
        "tcode_a": "FK01",  # Tworzenie dostawcy (Create Vendor)
        "tcode_b": "PAY01",  # Zatwierdzenie płatności (Approve Payment)
        "risk_level": "CRITICAL",
        "description": "Creating vendor and approving payment.",
    },
    {
        "risk_id": "SOD_02",
        "tcode_a": "FB01",  # Wpis księgowy (Post Entry)
        "tcode_b": "PAY01",  # Zatwierdzenie płatności (Approve Payment)
        "risk_level": "HIGH",
        "description": "Posting entries and executing payment run.",
    },
    {
        "risk_id": "SOD_03",
        "tcode_a": "ME21N",  # Zamówienie zakupu (Create PO)
        "tcode_b": "MIRO",  # Wprowadzenie faktury (Enter Invoice)
        "risk_level": "MEDIUM",
        "description": "Creating PO and entering invoice.",
    },
]

# Zapis w ramce danych p
df_sod = pd.DataFrame(sod_rules)

print("MACIERZ RYZYK SoD")
display(df_sod)
```
